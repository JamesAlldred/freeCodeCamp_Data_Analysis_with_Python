# Reading data

## pandas functions

## CSV, TXT

`pd.read_csv()`



Use `pd.read_csv?` to see the documentation for the function.

Can pass http or https url to read data from the web.

## SQL database

`pd.read_sql()`: Gets data from a SQL database and returns a DataFrame.
- `pd.to_sql()`: Writes data from a DataFrame to a SQL database.
    - If already exists --> error, can use `if_exists` parameter to specify what to do (replace, append, fail).

## HTML tables

`pd.read_html()`: Reads HTML table(s) into a list of DataFrame objects.
- Automtically detects headers from tags if html properly formatted
    - Needs cleaning if website not well formatted

## Excel files

`read_excel()`: Reads an Excel file into a pandas DataFrame.
- Can read specific sheets by name or index.
- Supports various Excel file formats (`.xls`, `.xlsx`, etc.).
- `read_excel()?` for documentation
- `to_excel()`: Writes a DataFrame to an Excel file.
    - Can specify sheet name, index, and other options.

`ExcelFile()`: Reads an Excel file into a pandas ExcelFile object, which can be used to read multiple sheets from the same file.
- `.sheet_names`: Returns a list of sheet names in the Excel file.
- `.parse(sheet_name)`: Reads a specific sheet from the Excel file into a DataFrame.
